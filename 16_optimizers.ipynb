### AdaGrad (Adaptive Gradient) :
* in this optimizer learning rate is not fixed, its adapted(change) according to the situation.
* if input features has different scales then this algorithm works better(but if it normalized then it is not useful)
* if features are sparse(most of the values are 0) then we can use this algorithm.
*** if data is sparse then its graph(contour plot) will be elongated(slope change at one axis and constant at other) at one dimension otherwise it will be some circular.
* such problems i.e slope change at one axis and constant at other is not abled to solved by BGD , momentum , NAG but by ADAGRAD.


**** What are the problems with these above algorit hms  :
