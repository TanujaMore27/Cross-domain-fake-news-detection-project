```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
datasets = ['ISOT', 'LIAR']

tfidf_scores = [0.99, 0.63]
sbert_scores = [0.94, 0.65]

x = np.arange(len(datasets))
width = 0.35

plt.figure()

plt.bar(x - width/2, tfidf_scores, width, label='TF-IDF')
plt.bar(x + width/2, sbert_scores, width, label='SBERT')

plt.xlabel("Dataset")
plt.ylabel("Accuracy")
plt.title("In-Domain Performance Comparison")

plt.xticks(x, datasets)
plt.legend()

plt.show()
```

```python
models = [
    "TF-IDF Logistic",
    "TF-IDF SVM",
    "SBERT Logistic"
]

accuracy = [0.51, 0.48, 0.55]

plt.figure()

plt.bar(models, accuracy)

plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.title("Cross-Domain Performance (Train: ISOT → Test: LIAR)")

plt.xticks(rotation=30)

plt.show()
```

#### Observations

* In-domain models achieve very high accuracy (up to 0.99 on ISOT).

* Performance on the LIAR dataset is lower due to shorter statements and different linguistic structure.

* Cross-domain experiments show a significant drop in accuracy (~0.48–0.55).

* SBERT embeddings slightly improve generalization compared to TF-IDF.
