### Cross-Domain Fake News Detection using TF-IDF

This notebook evaluates how well fake news detection models trained on
the ISOT dataset generalize to a different dataset, LIAR.

The experiment investigates the effect of domain shift in fake news
classification.

Training Dataset: ISOT (news articles)  
Testing Dataset: LIAR (political claims)

#### Experiment Setup

Training Data:
ISOT Dataset (news articles)

Testing Data:
LIAR Dataset (short political statements)

Feature Representation:
TF-IDF (Term Frequency–Inverse Document Frequency)

Models Used:
• Logistic Regression
• Support Vector Machine (SVM)

Goal:
Evaluate cross-domain generalization of fake news detection models.

#### Text Preprocessing

Text preprocessing is performed using spaCy with the following steps:

• Removing stopwords
• Removing punctuation
• Lemmatization

These steps help normalize the text before feature extraction.

```python
import spacy 
nlp = spacy.load('en_core_web_sm')
def preprocess(text):
    doc = nlp(text)
    filtered_tokens = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue
        filtered_tokens.append(token.lemma_)
    return " ".join(filtered_tokens)
```

```python
df['processed_text'] = df['Text'].apply(preprocess)
```

#### Observations

When trained and tested on the same dataset, models achieved very high
accuracy on the ISOT dataset.

However, when the models trained on ISOT are tested on the LIAR dataset,
performance drops significantly to around 48–51% accuracy.

This drop occurs because the two datasets have very different
characteristics:

• ISOT contains long news articles.
• LIAR contains short political claims.

As a result, vocabulary distribution, writing style, and sentence
structure differ significantly between the two datasets.

This experiment highlights the challenge of domain shift in fake news
detection.
